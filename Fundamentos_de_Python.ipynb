## Capitulo 3: Control de Flujo en Python

El control de flujo permite que un programa tome decisiones y repita acciones. En Python, esto se logra mediante estructuras condicionales y bucles.

### Estructuras condicionales: if, elif, else

Estas estructuras permiten ejecutar bloques de código basados en condiciones.

```python
edad = float(input("Ingresa tu edad: "))

if edad < 18:
    print("Eres menor de edad.")
elif edad == 18:
    print("Tienes 18 años.")
else:
    print("Eres mayor de edad.")
```

* Execute some code if a condition is met

![If-elseif](https://cdn.programiz.com/sites/tutorial2program/files/Python_if_elif_else_statement.jpg)

* You can add an "elif" (else if) and/or an "else"
  * elif: executes when the if-statement condition is not met
  * else: executes when the elif condition is not met

#### Opreaciones relacionales y lógicos

**a) Operadores relacionales**   
- Igual a (==)
- Diferente de (!=)
- Menor que (<)
- Mayor que (>)
- Menor o igual que (<=)
- Mayor o igual que (>=)

**b) Operadores lógicos**
- and: verdadero si ambas condiciones son verdaderas
- or: verdadero si al menos una condición es verdader
- not: Invierte el valor de verdad

```python
edad = float(input("Ingresa tu edad: "))
respuesta = input("Responde True (sí) o False (no); ¿Eres estudiante?: ").strip().lower() # Convertimos la respuesta del usuario en minúsculas para evitar errores

# Convertimos la cadena a un valor booleano
if respuesta == "true":
    estud = True
else:
    estud = False

if edad < 18 and estud==True :
    print("Eres menor de edad y estas estudiando.")
elif edad < 18 and estud == False:
    print("Eres menor de edad pero no estas estudiando.")
elif edad == 18 and estud==True:
    print("Tienes 18 años y estas estudiando.")
elif edad == 18 and estud==False:
    print("Tienes 18 años pero no estas estudiando.")
elif edad > 18 and estud==True:
    print("Eres mayor de edad y estas estudiando.")
else:
    print("Eres mayor de edad pero no estas estudiando.")
```

## Capítulo 4 Bucles (loops)

**a) for**
Se utiliza para iterar sobre una secuencia (lista, tupla, cadena, etc.).


```python
frutas = ["manzana", "banana", "cereza"]
for fruta in frutas:
    print("Me gusta la", fruta)
```

**Uso de range():**
La función range() genera una secuencia de números.

```python
for i in range(5):
    print(i)
```

**b) while**
Repite un bloque de código mientras una condición sea verdadera.
* Parecido al condicional if, but la maquina entra en un bucle, hasta que el condicional deja de cumplirse.
* Si la condicion establecida en el while-statement no se cumple el bucle es abandonado.

![while loop](https://www.tutorialspoint.com/python/images/python_while_loop.jpg)

```python
contador = 0
while contador < 5:
    print(contador)
    contador += 1
```

#### Bucles anidados
Es posible colocar un bucle dentro de otro. Esto es útil para trabajar con estructuras de datos multidimensionales o para generar combinaciones.

```python
for i in range(3):
    for j in range(2):
        print(f"i = {i}, j = {j}")
```

## Capítulo 5: Listas en python

Una lista es una estructura de datos en Python que permite almacenar múltiples elementos en una sola variable. Las listas son:
* Ordenadas: mantienen el orden de inserción de los elementos.
* Mutables: se pueden modificar después de su creación.
* Heterogéneas: pueden contener elementos de diferentes tipos de datos.
  - mi_lista = [10, "Hola", 3.14, True]

```python
lista_vacia = []
frutas = ["manzana", "banana", "cereza"]
print(frutas[0])  # Imprime "manzana"
print(frutas[-1])  # Imprime "cereza" (último elemento)
```

Cambiar el valor de un elemento

```python
frutas[1] = "kiwi"
print(frutas)
```

```python
frutas.append("naranja")  # Agrega al final
frutas.insert(1, "uva")  # Inserta en la posición 1
```

```python
print(frutas)
```

## Capítulo 6: Funciones en Python
Una función es un bloque de código que realiza una tarea específica y puede ser reutilizado en diferentes partes de un programa. Las funciones ayudan a:

* Dividir programas complejos en partes más manejables.

* Evitar la repetición de código.

* Mejorar la legibilidad y mantenimiento del código.

**Sintaxis para definir una función:**

```python
def nombre_de_la_funcion(parámetros):
    # Bloque de código
    return resultado  # Opcional
```

Ejemplo:


```python
def saludar(nombre):
    print(f"¡Hola, {nombre}!")

saludar("Ana")  # 'Ana' es el argumento
```

#### La instrucción return
La instrucción return se utiliza para devolver un valor desde una función al lugar donde fue llamada.

```python
def sumar(a, b):
    return a + b

resultado = sumar(3, 5)
print(resultado)  # Imprime 8
```

```python
# a function that returns the square of whatever number you input
def square(x):
  return x**2

# We can print a result of a function like so...
print(square(511))

# a function that substract the second number to the first you input
def restar(x, y):
  z = x - y
  return print(z)

# We can print a function just by calling it, if the function returns a print
restar(4865,3129)

# a functions that returns a "sum" of two string variables and a seperation
def stringConcat(str1, str2, sep):
  return str1 + sep + str2

#printing a string function
print(stringConcat("Hello","World!", " "))

# a function without a return statement, which returns None
def sumar(x,y):
  s = x + y
  return print(s)

resultado = sumar(3, 5)

# a function without a return, which also returns None
def producto(x,y):
  p = x*y
  return print(p)
```
