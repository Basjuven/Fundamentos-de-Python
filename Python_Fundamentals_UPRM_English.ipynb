## Chapter 3: Flow Control in Python

Flow control allows a program to make decisions and repeat actions. In Python, this is done using conditional structures and loops.

### Conditional Structures: `if`, `elif`, `else`

These structures allow us to execute blocks of code based on conditions.

```python
age = float(input("Enter your age: "))

if age < 18:
    print("You are a minor.")
elif age == 18:
    print("You are 18 years old.")
else:
    print("You are an adult.")
```

* Execute some code if a condition is met.

![If-elseif](https://cdn.programiz.com/sites/tutorial2program/files/Python_if_elif_else_statement.jpg)

* You can add an `elif` (`else if`) and/or an `else`.
  * `elif`: executes when the `if` statement condition is not met.
  * `else`: executes when the `elif` condition is not met.

#### Relational and Logical Operations

**a) Relational Operators**   
- Equal to (`==`)
- Not equal to (`!=`)
- Less than (`<`)
- Greater than (`>`)
- Less than or equal to (`<=`)
- Greater than or equal to (`>=`)

**b) Logical Operators**
- `and`: true if both conditions are true
- `or`: true if at least one condition is true
- `not`: reverses the truth value

```python
age = float(input("Enter your age: "))
answer = input("Answer True (yes) or False (no): Are you a student? ").strip().lower()  # Convert the user's answer to lowercase to avoid errors.

# Convert the string into a Boolean value.
if answer == "true":
    student = True
else:
    student = False

if age < 18 and student == True:
    print("You are a minor and you are studying.")
elif age < 18 and student == False:
    print("You are a minor, but you are not studying.")
elif age == 18 and student == True:
    print("You are 18 years old and you are studying.")
elif age == 18 and student == False:
    print("You are 18 years old, but you are not studying.")
elif age > 18 and student == True:
    print("You are an adult and you are studying.")
else:
    print("You are an adult, but you are not studying.")
```

## Chapter 4: Loops

**a) `for` loop**

A `for` loop is used to iterate over a sequence, such as a list, tuple, string, and so on.

```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print("I like", fruit)
```

**Using `range()`:**

The `range()` function generates a sequence of numbers.

```python
for i in range(5):
    print(i)
```

**b) `while` loop**

A `while` loop repeats a block of code as long as a condition is true.

* It is similar to an `if` conditional, but the computer enters a loop until the condition stops being true.
* If the condition established in the `while` statement is not met, the loop is exited.

![while loop](https://www.tutorialspoint.com/python/images/python_while_loop.jpg)

```python
counter = 0
while counter < 5:
    print(counter)
    counter += 1
```

#### Nested Loops

It is possible to place one loop inside another. This is useful when working with multidimensional data structures or generating combinations.

```python
for i in range(3):
    for j in range(2):
        print(f"i = {i}, j = {j}")
```

## Chapter 5: Lists in Python

A list is a data structure in Python that allows us to store multiple elements in a single variable. Lists are:

* Ordered: they preserve the insertion order of the elements.
* Mutable: they can be modified after they are created.
* Heterogeneous: they can contain elements of different data types.
  - `my_list = [10, "Hello", 3.14, True]`

```python
empty_list = []
fruits = ["apple", "banana", "cherry"]
print(fruits[0])   # Prints "apple"
print(fruits[-1])  # Prints "cherry" (last element)
```

Changing the Value of an Element

```python
fruits[1] = "kiwi"
print(fruits)
```

Adding an Element

```python
fruits.append("orange")  # Adds to the end
fruits.insert(1, "grape")  # Inserts at position 1
```

```python
print(fruits)
```

Removing an Element

```python
fruits.remove("grape")  # Removes "grape"
fruits.pop()  # Removes the last element
```

```python
print(fruits)
```

## Chapter 6: Functions in Python

A function is a block of code that performs a specific task and can be reused in different parts of a program. Functions help us to:

* Divide complex programs into more manageable parts.

* Avoid code repetition.

* Improve code readability and maintenance.

**Syntax for defining a function:**

```python
def function_name(parameters):
    # Code block
    return result  # Optional
```

Example:


```python
def greet(name):
    print(f"Hello, {name}!")

greet("Ana")  # 'Ana' is the argument
```

#### The `return` Statement

The `return` statement is used to send a value from a function back to the place where the function was called.

```python
def add(a, b):
    return a + b

result = add(3, 5)
print(result)  # Prints 8
```

```python
# A function that returns the square of any number you input.
def square(x):
  return x**2

# We can print the result of a function like this:
print(square(511))

# A function that subtracts the second number from the first number you input.
def subtract(x, y):
  z = x - y
  return print(z)

# We can print a function just by calling it if the function returns a print statement.
subtract(4865, 3129)

# A function that returns the concatenation of two string variables with a separator.
def string_concat(str1, str2, sep):
  return str1 + sep + str2

# Printing a string function.
print(string_concat("Hello", "World!", " "))

# A function with a return statement that returns the result of a print function.
def add_and_print(x, y):
  s = x + y
  return print(s)

result = add_and_print(3, 5)

# Another function that returns the result of a print function.
def product(x, y):
  p = x * y
  return print(p)
```
